```python
df['brand'] = df['model_name'].str.split().str[0]
```

```python
df['model'] = df['model_name'].str.replace(r'\(.*?\)', '', regex=True)
```

```python
df['color'] = df['model_name'].str.extract(r'\(([^,]+),')
```

```python


def clean_model_name(text):
    text = text.strip()

    # stop words ONLY for extra details (colors, charger, storage, etc.)
    stop_words = [
        "with", "and", "charger", "supervooc", "mah", "battery",
        "edition", "pulse", "green", "blue", "red", "black", "white",
        "camera", "storage", "locked", "sim", "power",
        "ram", "rom", "charger", "box","only", "on", "Jio","Typse","C","Charging","4000","6000mAh","HD"
    ]

    # stop at appearance of any extra descriptive word
    pattern = r"\b(" + "|".join(stop_words) + r")\b.*$"

    cleaned = re.sub(pattern, "", text, flags=re.IGNORECASE).strip()

    # remove trailing symbols like + -
    cleaned = re.sub(r"[\+\-]+$", "", cleaned).strip()

    return cleaned
```

```python
df['model']=df['model'].apply(clean_model_name)
```
